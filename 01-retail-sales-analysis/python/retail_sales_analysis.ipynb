# Retail Sales Analysis

My individual portfolio project. Fictional, synthetic data, January to March 2026, 40 records. Practice results, not real business performance.

Run the cells in order. Every number below is calculated from `data/retail_sales.csv`.

```python
import pandas as pd

df = pd.read_csv('data/retail_sales.csv', parse_dates=['order_date'])
print(df.shape)
print(df.head())
```

```python
# Data quality checks
print('Missing values:')
print(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
print('Revenue check errors:', (df['revenue_aud'] != df['unit_price_aud'] * df['quantity_sold']).sum())
print('Profit check errors:', (df['profit_aud'] != df['revenue_aud'] - df['cost_aud']).sum())
```

```python
# Overall performance
print('Revenue:', df['revenue_aud'].sum())
print('Cost:', df['cost_aud'].sum())
print('Profit:', df['profit_aud'].sum())
print('Units:', df['quantity_sold'].sum())
print('Margin %:', round(100 * df['profit_aud'].sum() / df['revenue_aud'].sum(), 1))
```

```python
# By category
cat = df.groupby('category').agg(revenue=('revenue_aud','sum'), profit=('profit_aud','sum'), units=('quantity_sold','sum'))
cat['margin_pct'] = round(100 * cat['profit'] / cat['revenue'], 1)
print(cat.sort_values('revenue', ascending=False))
```

```python
# By city
city = df.groupby('city').agg(revenue=('revenue_aud','sum'), profit=('profit_aud','sum'), units=('quantity_sold','sum'))
city['margin_pct'] = round(100 * city['profit'] / city['revenue'], 1)
print(city.sort_values('revenue', ascending=False))
```

```python
# Monthly trend
df['month'] = df['order_date'].dt.to_period('M').astype(str)
monthly = df.groupby('month').agg(revenue=('revenue_aud','sum'), profit=('profit_aud','sum'))
monthly['margin_pct'] = round(100 * monthly['profit'] / monthly['revenue'], 1)
print(monthly)
```

```python
# By product
prod = df.groupby('product').agg(revenue=('revenue_aud','sum'), profit=('profit_aud','sum'), units=('quantity_sold','sum'))
prod['margin_pct'] = round(100 * prod['profit'] / prod['revenue'], 1)
print(prod.sort_values('revenue', ascending=False))
```

```python
# Charts
import matplotlib.pyplot as plt

cat.sort_values('revenue', ascending=False).plot(kind='bar', y=['revenue','profit'], title='Revenue and profit by category')
plt.tight_layout(); plt.show()

monthly.plot(kind='line', marker='o', title='Monthly revenue and profit')
plt.tight_layout(); plt.show()
```

## What I found

- Electronics earns the most revenue (AUD 1,032), but Clothing has the highest margin (52.1%).
- Groceries sells the most units (75) but has the lowest revenue and margin.
- February is the peak month (AUD 1,218 revenue), March falls back slightly but stays above January.
